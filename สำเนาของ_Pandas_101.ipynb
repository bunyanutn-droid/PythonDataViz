### การวนลูป record ในตาราง  (```.iterrows```)
เป็นฟังก์ชันของ Pandas DataFrame ที่ใช้วนลูปทีละแถวของตารางในแต่ละรอบจะได้
* index = เลขแถว
* row = ข้อมูลของแถวนั้น (Series)
ซึ่งแตกต่างจาก for a in df['age'] ที่วนเฉพาะคอลัมน์เดียว ใช้เมื่อสนใจข้อมูลคอลัมน์เดียว

````
this_data = data_covid[['sex','age','province_of_onset']]
````

````
for each_row in this_data.iterrows(): #each_row เป็น Tuple(index, row)
    # print(each_row)เลขแถว
    # print(each_row[1]) ข้อมูลของแถว
    if (each_row[1]['age'] == 20) and (each_row[1]['province_of_onset'] == 'ขอนแก่น'):
      #ดึงค่าในคอลัมน์ age แล้วเช็คว่า=20มั้ย และดึงค่าจากคอลัมน์ province_of_onsetแล้วเช็คว่าเป็นขอนแก่นมั้ย
        print(each_row)
````

### การวนลูป แบบมองตารางแพนด้าส์(pandas dataframe) เป็น numpy array  (```.iloc```)
`.iloc` เป็นคำสั่งของ Pandas ที่ใช้เข้าถึงข้อมูล ตามตำแหน่งแถว (Row Index) และคอลัมน์ (Column Index) เหมือนการเข้าถึงอาร์เรย์ใน NumPy

````
for each_row in range(this_data.shape[0]):
    if (this_data.iloc[each_row,1] == 20) and (this_data.iloc[each_row,2] == 'ขอนแก่น'):  #each_row คือ ตัวเลข
        print(each_row)
        print(this_data.iloc[each_row,:])
````
